# Lowercasing

* Lowercasing is the process of converting all text into lowercase letters in NLP.
* Example: “Apple”, “APPLE”, and “apple” become “apple”.
* It is an important text preprocessing technique.
* It helps reduce the vocabulary size of the dataset.
* Removes unnecessary differences between words caused by capitalization.
* Makes text data more consistent and cleaner.
* Improves performance in tasks like:
    1. Text Classification
    2. Sentiment Analysis
    3. Tokenization
    4. Search Systems
* Helps machine learning models process text more efficiently.

# Removing HTML Tags

````python
import re 
def striphtml(data):
    p = re.compile(r'<.*?>')
    return p.sub('',data)
    
````

# Removing URL

````python
def remove_URL(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
````

````python
text = '''Hey everyone! Visit https://openai.com to explore the latest AI innovations and research updates. You can also learn Python programming from https://www.geeksforgeeks.org and practice coding problems on https://leetcode.com/problems.
For web development tutorials, check out https://developer.mozilla.org and https://www.w3schools.com/html/. Many students also watch free courses on https://youtube.com and read tech blogs from https://medium.com.
Follow AI trends on https://huggingface.co and explore datasets from https://kaggle.com/datasets. If you want cloud computing tutorials, visit https://aws.amazon.com or https://cloud.google.com. '''
````

````python
remove_URL(text)
````

````python
sample = '''Hii theree! 👋 I am curently lernning Natural Langauge Processing and text tokeniztion. It is very intresting becuase computers can understnd humen langauge using different algoritms and models. Somtimes the sentnces contain spelling erors, emojis 😂🔥, punctuations!!! and random symbls @#% which makes preprocessing more importent.Tokeniztion is the proces of spliting text into smaller parts like words, sentnces or subwords. For exampel, the sentnce “I love AI and machin learning” can be brokn into tokens such as ["I", "love", "AI", "and", "machin", "learning"].
Many NLP tasks like sentiment analyzis, text classifiction, spam detecton, and chatbots depend on good tokeniztion methods. Different libreries such as NLTK, spaCy, and HuggingFace provide powerfull tokenizers for handling complexx text data.Peoples often write text with incorret spellings, unicode charcters ❤★∞, emojis 🤖🚀😂, hashtags #AI #MachineLearning, and URLs like https://example.com which are useful for testing preprocessing pipelnes and cleaning functons.'''
````

````python
from nltk.tokenize import sent_tokenize,word_tokenize
sents = sent_tokenize(sample)
sents
````

````python
for sent in sents:
    print(word_tokenize(sent))
````
