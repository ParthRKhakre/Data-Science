# Removing HTML Tags

````python
import re
def striphtml(data):
    p = re.compile(r'<.*?>') # Regex pattern used to match HTML tags.
    return p.sub('',data)    # Replaces all match HTML tags with with empty string.
````

# Removing URL

````python
def remove_URL(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
````

# Remove Punctuation

````python
import string
import time
string.punctuation
````

````python
exclude = string.punctuation
````

````python
# logic coded from scratch hence the code is slow 
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,' ')
    return text    
````

````python
text = 'string. with .Punctuation?'
````

````python
start = time.time()

print(remove_punc(text))
time1 = time.time() - start 

print(time1)
````

````python
# We have another option for text translation
def remove_punc(text):
    return text.translate(str.maketrans('','',exclude))
````

````python
start = time.time()
print(remove_punc(text))
time2 = time.time() - start 

# translate code is faster around 18 times
````

# Chat Word Treatment

````python
chat_words = {
    "U": "you",
    "UR": "your",
    "BTW": "by the way",
    "IDK": "i don't know",
    "GR8": "great",
    "THX": "thanks",
    "OMG": "oh my god",
    "LOL": "laugh out loud",
    "BRB": "be right back",
    "TTYL": "talk to you later",
    "IMO": "in my opinion",
    "IMHO": "in my humble opinion",
    "FYI": "for your information",
    "ASAP": "as soon as possible",
    "PLZ": "please",
    "CU": "see you",
    "BCZ": "because",
    "B4": "before",
    "L8R": "later",
    "GM": "good morning",
    "GN": "good night",
    "TC": "take care",
    "NP": "no problem",
    "W8": "wait",
    "HBD": "happy birthday",
    "ROFL": "rolling on floor laughing",
    "ILY": "i love you"
}
````

````python

def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else: 
            new_text.append(w)
    return " ".join(new_text)            
````

````python
chat_conversion('IMHO he is the best')
````

# Removing Stopwords

````python
from nltk.corpus import stopwords
stopwords.words('english')
````

````python
def remove_stopwords(text):
    new_text = []
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append(' ')
        else :
            new_text.append(word)
    
    x = new_text[:]
    new_text.clear()
    return "".join(x)          
````

````python
text = '''
Natural Language Processing is a branch of artificial intelligence that helps 
computers understand human language. It is used in applications like chatbots, 
language translation, sentiment analysis, spam detection, and search engines. 
Many NLP preprocessing techniques include tokenization, stopword removal, stemming, 
lemmatization, and lowercasing. The goal of stopword removal is to remove commonly 
used words such as “is”, “the”, “and”, “of”, which do not add much meaning to the 
sentence. Students often practice NLP using Python libraries like NLTK and spaCy.'''
````

````python
remove_stopwords(text)
````

# UNICODE Text

````python
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags
        "\U00002702-\U000027B0"
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub(r'', text)
````
