# 3.Opening a CSV file from an URL

```python
import requests
from io import stringIO

url = ""
headers = {"User-Agent":"Mozilla/5.0(Macintosh;Intel Mac OS X 10.14;rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url,headers=headers)
data = stringIO(req.text)

pd.read_csv(data)
```

# 4. sep and names Parameter

```python
# .read_csv() by default have sep="," in tsv files 
# we need to override the parameter but column name 
# gets removed to add the name of columns use a parameter 
# called 'names' and pass a list
pd.read_csv("movie_titles_metadata.tsv",sep="\t",names=["sno","movie","year","rating","views","genre"])
```

# 5. index_col parameter

```python
# index_col parameter can be used to set any column 
# as index hence default pandas index would be replced 
# by the provided column.
pd.read_csv("aug_train.csv",index_col="enrollee_id")
```

# 6.Header parameter

```python
# Whenever you column is by mistake made a row pandas 
# Is unable to understand how to process it inorder to
# make the row as feature column use 'header'

pd.read_csv("test.csv",header=1)
```

# 7.usecols parameter

```python
# In ML you might only require a few parameter that you 
# need to use inorder to get that particular columns 
# 'usecols' is used.

pd.read_csv('aug_train.csv',usecols=['enrollee_id','gender','education_level'])
```

# 8.squeeze parameters

```python
# squeeze() is used to reduce dimensions commonly,
# Converts a DataFrame with one column into a Series.

pd.read_csv("aug_train.csv",usecols=['enrollee_id']).squeeze("columns")
```

# 9. skiprows/nrows Parameter

```python
# is used to skip a speicfic rows while reading a file,
# Indexing starts from 0 so make sure that 0 index is 
# features columns

pd.read_csv("aug_train.csv",skiprows=[1,3])
```

```python
# nrows is used to read only fixed number 
# of rows from a file.
pd.read_csv("aug_train.csv",nrows=100)
```

# 10. Encoding Parameter

```python
# Generally most dataset are UTF-8 but we might find 
# some dataset that are not encoded properly or encoded
# with some other encoding style
# UnicodeDecodeError is most common error in these case

pd.read_csv('zomato.csv',encoding='latin-1')
```

# 11.error_bad_lines / skip bad lines parameter

```python
# Skip rows with too many/few columns instead of throwing an error.
# when a parser error is received use error_bad_lines
pd.read_csv('zomato.csv',encoding='latin-1',on_bad_lines="skip")
```

# 12.dtypes parameter

```python
pd.read_csv("aug_train.csv",dtype={'target':int}).info()
```

# 13.Handling dates

```python
# Pandas treats objects as string 
# hence we cannot apply data-wise filter on it
# use 'parse-date' to convert a column to datetime
pd.read_csv("IPL Matches 2008-2020.csv",parse_dates=['date']).info()
```

# 14.convertors

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python
# converters is used to apply a custom function to column values while reading data.
pd.read_csv("IPL Matches 2008-2020.csv",converters={'team1':rename})
```

# 15. na_values parameters

```python
# na_values is used to tell pandas which values 
# should be treated as missing (NaN),generally 
# used to make place holders as NaN.

pd.read_csv("aug_train.csv",na_values=['Male',])
```
