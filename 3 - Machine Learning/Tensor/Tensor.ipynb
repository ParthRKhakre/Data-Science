Tensor - N dimensional array

2D - Matrix
3D - Tensors

[1,2,3,4] in programming it is known as vector,1D Array.

# Axes, Dimensions and Rank
> Axes = Dimensions = Ranks

* [1,2,3,4] is a 1D tensor but dimension of vector is 4D as there are four elements
* Here when four scaler where added together then 1D tensor was created
* In order to get the next dimension we need to added the previous dimensions 

1. vector = combination of multiple scaler
2. matrix = combinations of vectors
3. tensors = combination of matrix
