```python
import numpy as np 
import pandas as pd 

# Simple Linear Regression 
from sklearn.linear_model import LinearRegression

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Dependent variable (Exam Scores)
Y = np.array([50, 55, 65, 70, 80])

# Create and train the model
model = LinearRegression()
model.fit(X, Y)

# Parameters
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])

# Predict score for 6 hours of study
prediction = model.predict([[6]])
print("Predicted Score:", prediction[0])
```

# Linear Regression
* using Placement.csv dataset.

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("Placement.csv")
df.head(5)
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel("CGPA")
plt.ylabel("Package")
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
X 
```

```python
from sklearn.model_selection import train_test_split 
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)

# test_size = 0.2 means 20% data will be kept for testing and 80% will be used to train model 
# random_state is a seeding parameter ensuring reproducible results
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(X_train,y_train) 
# fit() is used to train the data
```

```python
print(X_test.head(5))
print(y_test.head(5))
```

```python
lr.predict(X_test.iloc[0].values.reshape(1,1))
```

```python
lr.predict(X_test.iloc[1].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color="red")
plt.xlabel("CG")
plt.ylabel("Package")
```
