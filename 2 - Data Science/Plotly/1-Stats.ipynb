# Measure of Central Tendency 


There are 3 Central Tendency 
1. Mean
2. Mode
3. Median

1.Mean - is the sum of values in the dataset divided by the number of values

> e.g. 3,1,2,4,5

mean(u) = (3 + 1 + 2 + 4 + 5) / 5 = 3



* Limitation of Mean : Outliers in data results into shifting in the mean Hence we need median

> 2.Median - the middle value in the dataset when data is arranged in order

e.g. 3,4,1,2,5,600
     1,2,3,4,5,600

median = 3.5

> 3.Mode - most frequent appearing value in the dataset It is specifically designed for categorical of discrete values

e.g. 1,1,2,3,1,4,5,6,1,2

mode = 1

# Graph for Univariate Analysis
type of Graph Depends on type of Data



> for such type of data we can plot Bar Chart


```python
import matplotlib.pyplot as plt

# Data
y = [60,40,30,35,20,15]
x = ['Beach','City','Adventure','Nature','Cruise','Others']

plt.bar(x,y)
```


>In relative frequency data Pie Chart is best suited

```python
import matplotlib.pyplot as plt

x = ['Beach','City','Adventure','Nature','Cruise','Others']
y = [60,40,30,35,20,15]
z = [0.3,0.2,0.15,0.175,0.1,0.075]

final_labels = [f"{l} ({e})" for l, e in zip(x, z)]

plt.pie(y,labels=final_labels,autopct='%1.1f%%')
```
